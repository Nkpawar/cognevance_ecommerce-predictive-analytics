```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries loaded successfully!")
```

```python
file_path = "../data/raw/online_retail_II.xlsx"

df = pd.read_excel(file_path)

print("Dataset loaded successfully!")
print("Rows:", len(df))
print("Columns:", len(df.columns))
```

```python
df.head()
```

```python
df.info()
```

```python
missing_values = df.isnull().sum()

missing_values
```

```python
missing_percentage = (
    df.isnull().sum() / len(df) * 100
).round(2)

missing_percentage
```

```python
duplicate_count = df.duplicated().sum()

print("Duplicate records:", duplicate_count)
```

```python
df.describe()
```

```python
print("Unique invoices:", df["Invoice"].nunique())
print("Unique products:", df["StockCode"].nunique())
print("Unique customers:", df["Customer ID"].nunique())
print("Unique countries:", df["Country"].nunique())
```

```python
df["InvoiceDate"] = pd.to_datetime(df["InvoiceDate"])

print("Earliest transaction:", df["InvoiceDate"].min())
print("Latest transaction:", df["InvoiceDate"].max())
```

```python
negative_quantity = (df["Quantity"] < 0).sum()

print("Transactions with negative quantity:", negative_quantity)
```

```python
print(
    "Percentage:",
    round(negative_quantity / len(df) * 100, 2),
    "%"
)
```

```python
invalid_price = (df["Price"] <= 0).sum()

print("Transactions with zero/negative price:", invalid_price)
```

```python
cancelled_invoices = df["Invoice"].astype(str).str.startswith("C").sum()

print("Cancelled transactions:", cancelled_invoices)
```
